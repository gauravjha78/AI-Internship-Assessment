```python
# Encoding with the following using Label Encoding.
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
encode_columns=['province', 'tourism_city', 'destination_type', 'season',
            'weather_condition', 'tourism_event_type', 'transport_mode_priority']

for col in encode_columns:
  df[col]=le.fit_transform(df[col]) # So first i got province so change the value of it to number like selecting the value one by one

print(df[['province']])
```

```python
le_demand=LabelEncoder()
le_category=LabelEncoder()
# Perform the operation on target column's
df['demand_category']=le_demand.fit_transform(df['demand_category'])
df['resource_allocation_decision']=le_category.fit_transform(df['resource_allocation_decision'])
```

```python
print(df['demand_category'])
print()
print("From here we start allocation")
print(df['resource_allocation_decision'])
```

```python
print("Demand_Classes:",le_demand.classes_)
print("Category_classes:",le_category.classes_)
```

```python
# Feature 1 - Revenue Per Tourist
df['revenue_per_tourist'] = df['tourism_revenue_yuan'] / df['tourist_arrivals']

# Feature 2 - International Tourist Ratio
df['international_ratio'] = df['international_tourists'] / df['tourist_arrivals']

# Feature 3 - Infrastructure Score
df['infra_score'] = (df['infrastructure_availability_score'] +
                     df['transport_service_score'] +
                     df['accommodation_capacity_score']) / 3

# Feature 4 - Revenue Efficiency (how much revenue per cost spent)
df['revenue_efficiency'] = df['tourism_revenue_yuan'] / df['operational_cost_yuan']

# Verify
print(df[['revenue_per_tourist', 'international_ratio', 'infra_score','revenue_efficiency']].head())
print("New shape:", df.shape)
```

```python
# Separate features and targets
X = df.drop(columns=['demand_category', 'resource_allocation_decision'])
y1=df['demand_category']
y2=df['resource_allocation_decision']

print("X shape:", X.shape)
print("y1 shape:", y1.shape)
print("y2 shape:", y2.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y1_train,y1_test,y2_train,y2_test=train_test_split(
    X,y1,y2,
    test_size=0.2,
    random_state=42,
    stratify=y1
)

print("Training size:", X_train.shape)
print("Testing size:", X_test.shape)
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# Function to train and evaluate
def train_eval(name, model, X_tr, y_tr, X_te, y_te):
    model.fit(X_tr, y_tr)
    pred = model.predict(X_te)
    acc = accuracy_score(y_te, pred)
    prec = precision_score(y_te, pred, average='weighted', zero_division=0)
    rec = recall_score(y_te, pred, average='weighted', zero_division=0)
    f1 = f1_score(y_te, pred, average='weighted', zero_division=0)
    print(f"{name} -> Acc: {acc:.4f} | Prec: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")
    return model, f1

# Model 1 - Baseline
m1, f1_1 = train_eval("Baseline", DummyClassifier(strategy='most_frequent'), X_train, y1_train, X_test, y1_test)

# Model 2 - Decision Tree
m2, f1_2 = train_eval("Decision Tree", DecisionTreeClassifier(max_depth=10, random_state=42), X_train, y1_train, X_test, y1_test)

# Model 3 - Random Forest
m3, f1_3 = train_eval("Random Forest", RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1), X_train, y1_train, X_test, y1_test)

# Model 4 - XGBoost
m4, f1_4 = train_eval("XGBoost", XGBClassifier(n_estimators=200, random_state=42, eval_metric='mlogloss'), X_train, y1_train, X_test, y1_test)

# Pick best model
models = {"Baseline": (m1, f1_1), "Decision Tree": (m2, f1_2), "Random Forest": (m3, f1_3), "XGBoost": (m4, f1_4)}
best_name = max(models, key=lambda x: models[x][1])
best_model_demand = models[best_name][0]
print(f"\nBEST MODEL: {best_name}")
```

```python
import joblib

joblib.dump(best_model_demand, 'model_demand.pkl')
print("Demand model saved!")
```

```python
# Now using y2 instead of y1 so i will know which is best
# Model 1 - Baseline
m1, f1_1 = train_eval("Baseline", DummyClassifier(strategy='most_frequent'), X_train, y2_train, X_test, y2_test)

# Model 2 - Decision Tree
m2, f1_2 = train_eval("Decision Tree", DecisionTreeClassifier(max_depth=10, random_state=42), X_train, y2_train, X_test, y2_test)

# Model 3 - Random Forest
m3, f1_3 = train_eval("Random Forest", RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1), X_train, y2_train, X_test, y2_test)

# Model 4 - XGBoost
m4, f1_4 = train_eval("XGBoost", XGBClassifier(n_estimators=200, random_state=42, eval_metric='mlogloss'), X_train, y2_train, X_test, y2_test)

# Pick best model
models = {"Baseline": (m1, f1_1), "Decision Tree": (m2, f1_2), "Random Forest": (m3, f1_3), "XGBoost": (m4, f1_4)}
best_name = max(models, key=lambda x: models[x][1])
best_model_resource = models[best_name][0]
print(f"\nBEST MODEL: {best_name}")

# Save it
joblib.dump(best_model_resource, 'model_resource.pkl')
print("Resource model saved!")
```

```python
importances = best_model_resource.feature_importances_
feat_imp = pd.Series(importances, index=X.columns).sort_values(ascending=False)

print("Top 10 most important features:")
print(feat_imp.head(10))

# Plot
plt.figure(figsize=(10,6))
feat_imp.head(15).sort_values().plot(kind='barh', color='steelblue')
plt.title('Top 15 Features - Resource Allocation Model')
plt.xlabel('Importance')
plt.tight_layout()
plt.savefig('feature_importance_resource.png')
plt.show()
```
